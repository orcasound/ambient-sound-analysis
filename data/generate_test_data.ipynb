## Download data for analysis
From 2024/09/15 00:00~2024/09/21 23:59

```python
delta_f = 1
bands = 12
delta_t = 1
loc = Hydrophone.ORCASOUND_LAB

pipeline = NoiseAnalysisPipeline(loc, 
                                delta_f=delta_f, bands=bands,
                                delta_t=delta_t, mode='safe')
```

```python
import collections
# generate data from 2024/09/16 to 2024/09/22
year = 2024
month = 9
start_day = 15
end_day = 21

for day in range(start_day, end_day + 1):
    # psd_dfs = []
    # broadband_dfs = []

    for hour in range(0, 24):   
        start_dt = dt.datetime(year, month, day, hour)
        end_dt = start_dt + dt.timedelta(hours=1)
        if start_dt < dt.datetime(2024, 9, 20, 11):
            continue
        datetime_str = f"{year}{month:02d}{day:02d}_{hour:02d}"
        print(f"Processing date: {datetime_str}")

        try:
            # generate parquet for this hour
            psd_path, broadband_path = pipeline.generate_parquet_file(start_dt, end_dt, upload_to_s3=False)
            
            # read hourly data
            psd_df = pd.read_parquet(psd_path)
            # psd_dfs.append(psd_df)

            broadband_df = pd.read_parquet(broadband_path)
            # broadband_dfs.append(broadband_df)

            #  write once per day
            psd_df.to_parquet(f"psd_{datetime_str}.parquet", index=False)
            broadband_df.to_parquet(f"broadband_{datetime_str}.parquet", index=False)
            
        except Exception as e:
            print(f"Error processing date {datetime_str}: {e}")
            continue




    # pd.concat(psd_dfs, axis=0).to_parquet(f"psd_{year}_{month:02d}_{day:02d}.parquet", index=False)
    # pd.concat(broadband_dfs, axis=0).to_parquet(f"broadband_{year}_{month:02d}_{day:02d}.parquet", index=False)
```

```python
import os
import pandas as pd

# Concatenate parquet files based on day
dates = ['20240915', '20240916', '20240917', '20240918', '20240919', '20240920', '20240921']

for date in dates:
    psd_dfs = []
    broadband_dfs = []
    print(f"Processing date: {date}")
    for hour in range(0, 24):
        datetime_str = f"{date}_{hour:02d}"
        psd_file = f"psd_{datetime_str}.parquet"
        broadband_file = f"broadband_{datetime_str}.parquet"
        
        if os.path.exists(psd_file):
            psd_df = pd.read_parquet(psd_file)
            psd_dfs.append(psd_df)
        
        if os.path.exists(broadband_file):
            broadband_df = pd.read_parquet(broadband_file)
            broadband_dfs.append(broadband_df)
    
    if psd_dfs:
        pd.concat(psd_dfs, axis=0).to_parquet(f"psd_{date}.parquet", index=False)
    
    if broadband_dfs:
        pd.concat(broadband_dfs, axis=0).to_parquet(f"broadband_{date}.parquet", index=False)
```
